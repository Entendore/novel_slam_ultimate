# Novel SLAM - Paper Figures Generator

This notebook generates publication-ready figures for your SLAM paper.

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from pathlib import Path

# Set publication-quality plot style
plt.style.use('seaborn-v0_8-paper')
plt.rcParams['figure.dpi'] = 300
plt.rcParams['font.size'] = 10
```

## 1. Load Trajectory Data

```python
# Load trajectory
traj_file = Path('../datasets/robot_1_trajectory.csv')
if traj_file.exists():
    df = pd.read_csv(traj_file)
    print(f'Loaded {len(df)} trajectory points')
    df.head()
```

## 2. Plot Trajectory

```python
if traj_file.exists():
    fig, ax = plt.subplots(figsize=(8, 8))
    ax.plot(df['x'], df['y'], 'b-', linewidth=2, label='Estimated')
    ax.set_xlabel('X (m)')
    ax.set_ylabel('Y (m)')
    ax.set_title('SLAM Trajectory')
    ax.legend()
    ax.grid(True)
    ax.axis('equal')
    plt.tight_layout()
    plt.savefig('../results/trajectory_plot.png', dpi=300, bbox_inches='tight')
    plt.show()
```

## 3. Plot Performance Metrics

```python
if traj_file.exists():
    fig, axes = plt.subplots(2, 2, figsize=(12, 10))
    
    # FPS
    axes[0, 0].plot(df['time'], df['fps'], 'g-')
    axes[0, 0].set_title('Processing Speed')
    axes[0, 0].set_xlabel('Time (s)')
    axes[0, 0].set_ylabel('FPS')
    axes[0, 0].grid(True)
    
    # Computation time
    axes[0, 1].plot(df['time'], df['comp_time_ms'], 'r-')
    axes[0, 1].set_title('Computation Time')
    axes[0, 1].set_xlabel('Time (s)')
    axes[0, 1].set_ylabel('ms')
    axes[0, 1].grid(True)
    
    # Uncertainty
    axes[1, 0].plot(df['time'], df['cov_x'], label='X')
    axes[1, 0].plot(df['time'], df['cov_y'], label='Y')
    axes[1, 0].plot(df['time'], df['cov_theta'], label='Theta')
    axes[1, 0].set_title('Pose Uncertainty')
    axes[1, 0].set_xlabel('Time (s)')
    axes[1, 0].set_ylabel('Variance')
    axes[1, 0].legend()
    axes[1, 0].grid(True)
    
    # Loop closures
    loop_mask = df['loop_closure'] == 'True'
    axes[1, 1].scatter(df.loc[loop_mask, 'x'], df.loc[loop_mask, 'y'], c='red', s=50, label='Loop Closures')
    axes[1, 1].plot(df['x'], df['y'], 'b-', alpha=0.5)
    axes[1, 1].set_title('Loop Closures')
    axes[1, 1].set_xlabel('X (m)')
    axes[1, 1].set_ylabel('Y (m)')
    axes[1, 1].legend()
    axes[1, 1].grid(True)
    axes[1, 1].axis('equal')
    
    plt.tight_layout()
    plt.savefig('../results/performance_metrics.png', dpi=300, bbox_inches='tight')
    plt.show()
```

## 4. Summary Statistics

```python
if traj_file.exists():
    print('=== SLAM Performance Summary ===')
    print(f'Total time: {df["time"].max():.2f} seconds')
    print(f'Average FPS: {df["fps"].mean():.2f}')
    print(f'Average computation time: {df["comp_time_ms"].mean():.2f} ms')
    print(f'Loop closures detected: {loop_mask.sum()}')
    print(f'Final uncertainty (X): {df["cov_x"].iloc[-1]:.6f}')
    print(f'Final uncertainty (Y): {df["cov_y"].iloc[-1]:.6f}')
```
